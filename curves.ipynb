```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 创建数据网格
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

# 定义几个不同的函数
Z1 = np.exp(-(X**2 + Y**2))  # 高斯函数
Z2 = np.sin(X) * np.cos(Y)   # 正弦余弦函数
Z3 = X**2 - Y**2             # 鞍面

# 创建子图
fig = plt.figure(figsize=(15, 5))

# 第一个子图
ax1 = fig.add_subplot(131, projection='3d')
surf1 = ax1.plot_surface(X, Y, Z1, cmap='plasma')
ax1.set_title('高斯函数')

# 第二个子图
ax2 = fig.add_subplot(132, projection='3d')
surf2 = ax2.plot_surface(X, Y, Z2, cmap='coolwarm')
ax2.set_title('正弦余弦函数')

# 第三个子图
ax3 = fig.add_subplot(133, projection='3d')
surf3 = ax3.plot_surface(X, Y, Z3, cmap='viridis')
ax3.set_title('鞍面')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.interpolate import griddata
from scipy.ndimage import gaussian_filter

# 原始矩阵
matrix = np.random.rand(100, 100)

# 方法1: 使用高斯滤波平滑
matrix_smooth = gaussian_filter(matrix, sigma=2)  # sigma控制平滑程度

# 方法2: 使用更高分辨率的插值
# 创建更密集的网格
x_original = np.linspace(0, 10, matrix.shape[1])
y_original = np.linspace(0, 10, matrix.shape[0])
X_original, Y_original = np.meshgrid(x_original, y_original)

# 创建更密集的目标网格
x_dense = np.linspace(0, 10, 200)  # 增加到200个点
y_dense = np.linspace(0, 10, 200)
X_dense, Y_dense = np.meshgrid(x_dense, y_dense)

# 将原始数据点展平用于插值
points = np.column_stack((X_original.ravel(), Y_original.ravel()))
values = matrix_smooth.ravel()

# 使用三次插值
matrix_interpolated = griddata(points, values, (X_dense, Y_dense), method='cubic')

# 绘制对比图
fig = plt.figure(figsize=(15, 5))

# 原始数据
ax1 = fig.add_subplot(131, projection='3d')
x_range = np.linspace(0, 10, matrix.shape[1])
y_range = np.linspace(0, 10, matrix.shape[0])
X, Y = np.meshgrid(x_range, y_range)
surf1 = ax1.plot_surface(X, Y, matrix, cmap='plasma', alpha=0.8)
ax1.set_title('原始数据')

# 高斯平滑后
ax2 = fig.add_subplot(132, projection='3d')
surf2 = ax2.plot_surface(X, Y, matrix_smooth, cmap='plasma', alpha=0.8)
ax2.set_title('高斯平滑')

# 插值平滑后
ax3 = fig.add_subplot(133, projection='3d')
surf3 = ax3.plot_surface(X_dense, Y_dense, matrix_interpolated, cmap='plasma', alpha=0.8)
ax3.set_title('插值平滑')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 示例矩阵
matrix = np.random.rand(100, 100)

# 自定义x, y坐标范围
x_range = np.linspace(0, 10, matrix.shape[1])
y_range = np.linspace(0, 10, matrix.shape[0])
X, Y = np.meshgrid(x_range, y_range)

# 绘制 - 使用参数优化平滑度
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X, Y, matrix, 
                      cmap='plasma', 
                      alpha=0.8,
                      linewidth=0,        # 去掉网格线
                      antialiased=True,   # 开启抗锯齿
                      rcount=200,         # 增加行渲染数量
                      ccount=200,         # 增加列渲染数量
                      shade=True)         # 开启阴影

ax.set_xlabel('X坐标')
ax.set_ylabel('Y坐标')
ax.set_zlabel('矩阵值')
ax.set_title('优化渲染参数的平滑曲面')

fig.colorbar(surf)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def moving_average_2d(matrix, window_size=3):
    """2D移动平均平滑"""
    from scipy.ndimage import uniform_filter
    return uniform_filter(matrix, size=window_size)

# 原始矩阵
matrix = np.random.rand(100, 100)

# 应用移动平均
matrix_smooth = moving_average_2d(matrix, window_size=5)

# 绘制
x_range = np.linspace(0, 10, matrix.shape[1])
y_range = np.linspace(0, 10, matrix.shape[0])
X, Y = np.meshgrid(x_range, y_range)

fig = plt.figure(figsize=(12, 5))

# 原始数据
ax1 = fig.add_subplot(121, projection='3d')
surf1 = ax1.plot_surface(X, Y, matrix, cmap='plasma', alpha=0.8, linewidth=0)
ax1.set_title('原始数据')

# 平滑后数据
ax2 = fig.add_subplot(122, projection='3d')
surf2 = ax2.plot_surface(X, Y, matrix_smooth, cmap='plasma', alpha=0.8, linewidth=0)
ax2.set_title('移动平均平滑')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.ndimage import gaussian_filter

def create_smooth_surface(matrix, smooth_factor=2, resolution_factor=2):
    """
    创建平滑的3D曲面
    
    Parameters:
    matrix: 输入矩阵
    smooth_factor: 平滑程度 (sigma for gaussian filter)
    resolution_factor: 分辨率提升倍数
    """
    
    # 1. 首先对数据进行高斯平滑
    matrix_smooth = gaussian_filter(matrix, sigma=smooth_factor)
    
    # 2. 创建更高分辨率的网格
    original_x = np.linspace(0, 10, matrix.shape[1])
    original_y = np.linspace(0, 10, matrix.shape[0])
    
    new_size_x = matrix.shape[1] * resolution_factor
    new_size_y = matrix.shape[0] * resolution_factor
    
    # 3. 使用插值创建高分辨率数据
    from scipy.interpolate import RectBivariateSpline
    
    spline = RectBivariateSpline(
        np.linspace(0, 10, matrix.shape[0]),
        np.linspace(0, 10, matrix.shape[1]),
        matrix_smooth,
        kx=3, ky=3  # 三次样条插值
    )
    
    x_new = np.linspace(0, 10, new_size_x)
    y_new = np.linspace(0, 10, new_size_y)
    X_new, Y_new = np.meshgrid(x_new, y_new)
    
    Z_new = spline(y_new, x_new)
    
    return X_new, Y_new, Z_new

# 原始矩阵
matrix = np.random.rand(100, 100)  # 使用较小的矩阵以便更好地看到效果

# 创建平滑曲面
X_smooth, Y_smooth, Z_smooth = create_smooth_surface(matrix, smooth_factor=1.5, resolution_factor=3)

# 原始坐标
x_range = np.linspace(0, 10, matrix.shape[1])
y_range = np.linspace(0, 10, matrix.shape[0])
X_original, Y_original = np.meshgrid(x_range, y_range)

# 绘制对比
fig = plt.figure(figsize=(15, 6))

# 原始曲面
ax1 = fig.add_subplot(121, projection='3d')
surf1 = ax1.plot_surface(X_original, Y_original, matrix, 
                        cmap='plasma', alpha=0.8)
ax1.set_title('原始曲面')
ax1.set_xlabel('X坐标')
ax1.set_ylabel('Y坐标')
ax1.set_zlabel('矩阵值')

# 平滑曲面
ax2 = fig.add_subplot(122, projection='3d')
surf2 = ax2.plot_surface(X_smooth, Y_smooth, Z_smooth, 
                        cmap='plasma', 
                        alpha=0.8,
                        linewidth=0,
                        antialiased=True,
                        shade=True)
ax2.set_title('平滑曲面')
ax2.set_xlabel('X坐标')
ax2.set_ylabel('Y坐标')
ax2.set_zlabel('矩阵值')

plt.tight_layout()
plt.show()
```
